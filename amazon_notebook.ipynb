#### A description for each of the columns in the dataset:

1. **order_id**: A unique identifier for each order placed.
2. **order_date**: The date and time when the order was placed.
3. **status**: The current status of the order (e.g., processing, shipped, delivered).
4. **item_id**: Unique identifier for each item in the order.
5. **sku**: Stock Keeping Unit, a unique code assigned to each distinct product for inventory management.
6. **qty_ordered**: The quantity of the specific item ordered.
7. **price**: The price per unit of the item.
8. **value**: The total value of the item in the order (price * quantity).
9. **discount_amount**: The amount of discount applied to the item.
10. **total**: The total cost of the item after discount.
11. **category**: The category to which the item belongs (e.g., electronics, clothing, groceries).
12. **payment_method**: The method used for payment (e.g., credit card, PayPal, cash on delivery).
13. **bi_st**: Business Intelligence status, indicating whether the order data has been processed for analytics.
14. **cust_id**: Unique identifier for each customer.
15. **year**: The year when the order was placed.
16. **month**: The month when the order was placed.
17. **ref_num**: Reference number associated with the order.
18. **Name Prefix**: Prefix added before a person's name (e.g., Mr., Mrs., Dr.).
19. **First Name**: The first name of the customer.
20. **Middle Initial**: The middle initial of the customer's name.
21. **Last Name**: The last name of the customer.
22. **Gender**: The gender of the customer.
23. **age**: The age of the customer.
24. **full_name**: The full name of the customer (concatenation of first name, middle initial, and last name).
25. **E Mail**: The email address of the customer.
26. **Sign in date**: The date when the customer signed up or created an account.
27. **Phone No.**: The phone number of the customer.
28. **Place Name**: The name of the place associated with the customer (e.g., residence city).
29. **County**: The county where the customer resides.
30. **City**: The city where the customer resides.
31. **State**: The state where the customer resides.
32. **Zip**: The ZIP or postal code of the customer's location.
33. **Region**: The region where the customer resides.
34. **User Name**: The username associated with the customer's account.
35. **Discount_Percent**: The percentage of discount applied to the order.
